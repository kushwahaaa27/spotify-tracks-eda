# 🎵 Spotify Tracks EDA

Exploring **114,000 Spotify tracks** across **114 genres** and **31,000+ artists** — analyzing audio features like danceability, energy, and valence to uncover what makes a track hit.

**Starting Date :** 23 June 2026 

```python
# Importing Libraries !! 

import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

#Setting the plot style

plt.style.use("seaborn-v0_8")
```

```python
#Loading the Dataset 
df = pd.read_csv("spotify-tracks-dataset-detailed.csv")
```

# 🧹 Data Cleaning 🧹

```python
df.isnull().sum()
```

```python
df["artists"] = df["artists"].fillna("Unknown")
df["album_name"] = df["album_name"].fillna("Unknown")
df["track_name"] = df["track_name"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

# 🔍 Exploratory Data Analysis ✨

```python
plt.figure(figsize=(8,5))
df["popularity"].plot(kind="hist",color = "coral" , edgecolor = "black", bins =10)
plt.title("Popularity of Songs",fontsize=15,fontstyle="italic",fontweight="bold")
plt.xlabel("Popularity Rankings")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["track_genre"].value_counts().head(10).plot(kind="bar",color="seagreen",edgecolor="black")
plt.title("Track Count by Genre (Balanced Dataset)", fontsize=15 ,fontstyle="italic",fontweight="bold")
plt.xlabel("Genres")
plt.xticks(rotation=45)
plt.ylabel("Track Count")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
df["explicit"].map({True:"Explicit",False:"Non-Explicit"}).value_counts().plot(kind="pie",autopct="%1.1f%%")
plt.title("Explicit vs Non-Explicit tracks",fontsize=15,fontstyle="italic",fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
features = ["danceability", "energy", "speechiness", "acousticness", 
            "instrumentalness", "liveness", "valence", "tempo"]

fig, axes = plt.subplots(2,4,figsize=(16,8))
axes = axes.flatten()

for i,feature in enumerate(features) : 

    df[feature].plot(kind="hist",color="steelblue",edgecolor="black" , ax=axes[i])
    axes[i].set_title(feature,fontsize=12,fontstyle="italic",fontweight="bold")
    axes[i].set_xlabel(feature)

plt.suptitle("Audio Features Distribution",fontsize=15,fontstyle="italic",fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Audio Features", fontsize=15, fontweight="bold", fontstyle="italic")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.groupby("track_genre")["popularity"].mean().sort_values(ascending=False).head(10).plot(kind="barh",color="darkorange",edgecolor="black",linewidth=1.5)
plt.title("Top 10 genres by average popularity",fontsize=15,fontstyle="italic",fontweight="bold")
plt.ylabel("Genre")
plt.xlabel("Average Popularity")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.plot(kind="scatter", x="loudness", y="energy", color="tan", alpha=0.08)
plt.title(" Loudness vs Energy scatter plot",fontsize=16,fontstyle="italic",fontweight="bold")
plt.xlabel("Loudness")
plt.ylabel("Energy")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.plot(kind="scatter",x="valence",y="popularity",alpha=0.01,color="coral")
plt.title(" Valence vs Popularity scatter plot",fontsize=15,fontstyle="italic",fontweight="bold")
plt.xlabel("Valence")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

```python
df.drop_duplicates(subset="track_name",inplace=True)
df[["track_name","popularity"]].sort_values(by="popularity",ascending=True).set_index("track_name").tail(10)\
    .plot(kind="barh",color="crimson",edgecolor="black",linewidth=1.5,figsize=(10,5),legend=False)
plt.title("Top 10 Most Popular Tracks", fontsize=15, fontweight="bold", fontstyle="italic")
plt.xlabel("Popularity")
plt.ylabel("Track Name")
plt.tight_layout()
plt.show()
```

```python
sep_artist = [] 

for artist in df["artists"] : 

    sep_artist.extend([c.strip() for c in artist.split(";")])

artist_split = pd.Series(sep_artist)

plt.figure(figsize=(8,5))
artist_split.value_counts(ascending=True).tail(10).plot(kind="barh",edgecolor="black",linewidth=1.5)
plt.title("Top 10 artists by track count", fontsize=15,fontstyle="italic",fontweight="bold")
plt.ylabel("Artists")
plt.xlabel("Count")
plt.tight_layout()
plt.show()
```

```python
df.boxplot(column="popularity", by="explicit", figsize=(10, 6))
plt.title("Explicit vs Popularity", fontsize=15, fontweight="bold", fontstyle="italic")
plt.suptitle("")  # to remove the ugly default title that pandas adds
plt.ylabel("Popularity")
plt.xlabel("")
plt.xticks([1, 2], ["Non-Explicit", "Explicit"])
plt.tight_layout()
plt.show()
```

```python
df["duration_min"] = df["duration_ms"]*(1/60000)

plt.figure(figsize=(8,5))
plt.hist(df["duration_min"],bins=100,color="crimson",edgecolor="black",linewidth=1.6)
plt.title("Duration distribution",fontsize=16,fontstyle="italic",fontweight="bold")
plt.xlabel("Duration")
plt.xlim(0, 20)
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
print("The Average Song Duration On Spotify Is : ",round(df["duration_min"].mean(),2) , "minutes")
```

```python
key_map = {0:"C", 1:"C#", 2:"D", 3:"D#", 4:"E", 5:"F", 
           6:"F#", 7:"G", 8:"G#", 9:"A", 10:"A#", 11:"B"}
df["key_name"] = df["key"].map(key_map)


print("")
print("🎵 Key Distribution\n")

print("In music, a key defines the home base note a song is built around — like C, G, or A.")
print("This plot reveals which musical keys are most commonly used across 114k Spotify tracks.")
print("")

df["key_name"].value_counts().plot(kind="bar",color="seagreen",edgecolor="black",linewidth=1.5,figsize=(8,5))
plt.title("Key distribution (which musical key is most common)",fontsize=12,fontstyle="italic",fontweight="bold")
plt.xlabel("Keys")
plt.xticks(rotation=0)
plt.ylabel("Count")
plt.tight_layout()
plt.show()
           
```

```python
df["mode_map"]=df["mode"].map({0:"Minor",1:"Major"})

print("")
print("Major vs Minor Mode Distribution\n")
print("In music, Major keys sound happy and upbeat while Minor keys sound sad and emotional.")
print("This plot shows how many tracks in the dataset fall into each category.\n")

df["mode_map"].value_counts().plot(kind="bar",color="teal",edgecolor="black",linewidth=1.5,figsize=(8,5))
plt.title("Major vs Minor mode distribution",fontsize=15,fontstyle="italic",fontweight="bold")
plt.xlabel("Mode")
plt.xticks(rotation=0)
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
df.plot(kind="scatter",x="acousticness",y="energy",color="pink",alpha=0.01,figsize=(8,5))
plt.title("Acousticness vs Energy : Scatter Plot",fontsize=16,fontstyle="italic",fontweight="bold")
plt.ylabel("Energy")
plt.xlabel("Acousticness")
plt.tight_layout()
plt.show()
```

```python
df.groupby("track_genre")["danceability"].mean().sort_values(ascending=True).tail(10)\
    .plot(kind="barh",edgecolor="black",linewidth=1.5,figsize=(8,5))
plt.title("Top 10 genres by danceability",fontsize=16,fontstyle="italic",fontweight="bold")
plt.xlabel("Average Daceability Score")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

## 📝 Conclusion

This EDA explored **113,550 Spotify tracks** across **114 genres** and **31,000+ artists** after cleaning.

### 🎵 Key Findings

- **Pritam** is the most prolific artist in the dataset — highlighting the strong presence of Bollywood music
- **The Beatles & George Jones** dominate in track count among Western artists
- **G Major** is the most commonly used musical key — likely due to its popularity in guitar-based music
- **Major key** songs outnumber Minor key songs — meaning more happy/upbeat tracks than sad ones
- **Explicit tracks** are slightly more popular on average than non-explicit ones — but both can hit high popularity
- **Energy and Loudness** show a strong positive correlation — louder tracks are almost always more energetic
- **Acousticness and Energy** show a clear negative correlation — acoustic tracks tend to be low energy
- **Valence and Popularity** have no clear relationship — a happy sounding song isn't guaranteed to be popular
- Most tracks fall between **2–4 minutes** in duration — the sweet spot for mainstream music
- **Latin and dancehall** genres top the danceability chart — built for the dance floor
- All genres have exactly **1000 tracks** — this is a balanced dataset, not a real-world distribution
